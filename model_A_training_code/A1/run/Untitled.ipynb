```python
from pathlib import Path
import yaml

a1_dir = Path("/data/recycle_dataset/notebooks/model A/runs/recycle_yolo11n_14class_v1")
args_path = a1_dir / "args.yaml"

args = yaml.safe_load(args_path.read_text())
print("lr0:", args.get("lr0"))
print("data:", args.get("data"))
print("imgsz:", args.get("imgsz"))
print("batch:", args.get("batch"))
print("optimizer:", args.get("optimizer"))
```

```python
from ultralytics import YOLO

best_pt = a1_dir / "weights" / "best.pt"
data_yaml = args["data"]

a1_best = YOLO(str(best_pt))
a1_result = a1_best.val(data=data_yaml, imgsz=args.get("imgsz", 640))
print(a1_result.results_dict)
```
